```python
houses = [{"size":1200,"bedrooms":3,"price":250000},
          {"size":900,"bedrooms":2,"price":180000},
          {"size":1500,"bedrooms":4,"price":320000}
]
for house in houses:
    print(house["price"])

```

```python
def price_per_sqft(house):
    return house["price"]/house["size"]
print(price_per_sqft(houses[0]))
print(price_per_sqft(houses[1]))
```

```python
def is_affordable(house, budget):
    if house["price"]<=budget:
        return True
    else:
        return False

print(is_affordable(houses[1],200000))
```

```python
def filter_affordable(houses,budget):
    result=[]
    for house in houses:
        if is_affordable(house,budget):
            result.append(house)
    return result
affordable_houses = filter_affordable(houses,200000)
print(affordable_houses)
```

```python
def filter_by_bedrooms(houses,min_bedrooms):
    result=[]
    for house in houses:
        if house["bedrooms"]>=min_bedrooms:
            result.append(house)
    return result
affordable_bedrooms = filter_by_bedrooms(houses,4)
print(affordable_bedrooms)
```
