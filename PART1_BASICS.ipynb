```python
def filter_affordable(houses,budget):
    result=[]
    for house in houses:
        if is_affordable(house,budget):
            result.append(house)
    return result
affordable_houses = filter_affordable(houses,200000)
print(affordable_houses)
```

```python
# While loop
houses = [{"size":1200,"bedrooms":3,"price":250000},
          {"size":900,"bedrooms":2,"price":180000},
          {"size":1500,"bedrooms":4,"price":320000}]
budget = 1000000
total_spent = 0
i = 0
while total_spent<budget and i<len(houses):
    total_spent+=houses[i]["price"]
    i+=1
print(f"Bought {i} houses, spent {total_spent}")
# enumerate() gives you BOTH the index and the item, together, in each loop step.
# Without it, you'd have to manually track a counter like the "i" above.
for index, house in enumerate(houses):
    print(f"House #{index}: price = {house['price']}")
```

```python
loaded_houses = []
with open("houses.txt", "r") as f:
    for line in f:
        try:
            size, bedrooms, price = line.strip().split(",")
            loaded_houses.append({
                "size_sqft": int(size),
                "bedrooms": int(bedrooms),
                "price": int(price)
            })
        except ValueError:
            print(f"Warning: skipping malformed line: {line.strip()}")

print(loaded_houses)
```

```python
# class defines a new type of object - a blueprint
class House:
    # __init__ runs automatically whenever you create a new House.
    # "self" refers to "this specific house instance" - it's how the object.
    # keeps track of its own data, seperate from every other house object.
    def __init__(self, size_sqft, bedrooms, price):
        self.size_sqft = size_sqft
        self.bedrooms = bedrooms
        self.price = price
    # A method - a function that belongs to the class, and can use the self's data.
    def price_per_sqft(self):
        return self.price/self.size_sqft
    # Creating ("instantiating") an actual house object from the blueprint
h1 = House(1200, 3, 250000)
    # Notice: no dict brackets. You access data with a dot, and call methods like this:
print(h1.price)
print(h1.price_per_sqft())
```

```python
class House1:
    def __init__(self, size, bedrooms, price):
        self.size = size
        self.bedrooms = bedrooms
        self.price = price
    def is_affordable(self, budget):
        if self.price>budget:
            return False
        elif self.price<budget:
            return True
h2 = House1(1500, 4, 500000)
print(h2.price)
print(h2.is_affordable(600000))
```
